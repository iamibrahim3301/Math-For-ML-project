```python
# Normalize the data set
def mean(X):
  return np.sum(X)/len(X)


def varience(X):
  mean_value = mean(X)
  return np.sum((X-mean_value)**2)/len(X)


def norm(X):
  mean_value = mean(X)
  varience_value = varience(X)
  return (X-mean_value)/np.sqrt(varience_value)
```

```python
X_norm = norm(X)
Y_norm = norm(Y)
```
