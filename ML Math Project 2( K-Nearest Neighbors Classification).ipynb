```python
import numpy as np
import matplotlib.pyplot as plt

from collections import Counter
```

```python
points ={"blue":[[2,4], [1,3], [2,3], [2,2], [2,1]],
        "red": [[5,6], [4,5], [4,6], [6,6], [5,4]]}
```

```python
new_points = [3,3]
```

```python
def euclidean_distance(new_points, points):
  return np.sqrt(np.sum((np.array(points) - np.array(new_points)) ** 2))
```

```python
class KNearestNeighbors:
  def __init__(self, k=3):
    self.k = k
    self.point = None

  def fit(self,points):
    self.points = points

  def predict(self, new_points):
    distances = []

    for group in self.points:
      for feature in self.points[group]:
        distance = euclidean_distance(new_points, feature)
        distances.append([distance, group])



    catagories = [category[1] for category in sorted(distances)[:self.k]]
    return Counter(catagories).most_common(1)[0][0]
```

```python
clf = KNearestNeighbors(k=3)
clf.fit(points)
print(clf.predict(new_points))
```

```python
#Visualize

ax = plt.subplot()
ax.grid(True, color="#323232")
ax.set_facecolor("black")
ax.figure.set_facecolor("#121212")
ax.tick_params(axis="x", colors="white")
ax.tick_params(axis="y", colors="white")


for point in points["blue"]:
  ax.scatter(point[0], point[1], color="#104DCA", s=60)
for point in points["red"]:
  ax.scatter(point[0], point[1], color="#FF0000", s=60)



new_class = clf.predict(new_points)
color = "#FF0000" if new_class == "red" else "#104DCA"
ax.scatter(new_points[0], new_points[1], color=color, marker="*", s=200, zorder=100)


for point in points["blue"]:
  ax.plot([point[0], new_points[0]], [point[1], new_points[1]], color="#104DCA", linestyle="--")
for point in points["red"]:
  ax.plot([point[0], new_points[0]], [point[1], new_points[1]], color="#FF0000", linestyle="--")

plt.show()
```

```python
# 3D visualization
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter


points ={"blue":[[2,4,4], [1,3,5], [2,3.3,0], [2,2,1], [2,1,4]],
        "red": [[5,6,4], [4,5,7], [4,6,7], [6,6,6], [5,4,9]]}
new_points = [3,3,4]


def euclidean_distance(new_points, points):
  return np.sqrt(np.sum((np.array(points) - np.array(new_points)) ** 2))



class KNearestNeighbors:
  def __init__(self, k=3):
    self.k = k
    self.point = None

  def fit(self,points):
    self.points = points

  def predict(self, new_points):
    distances = []

    for group in self.points:
      for feature in self.points[group]:
        distance = euclidean_distance(new_points, feature)
        distances.append([distance, group])



    catagories = [category[1] for category in sorted(distances)[:self.k]]
    return Counter(catagories).most_common(1)[0][0]


clf = KNearestNeighbors(k=3)
clf.fit(points)
print(clf.predict(new_points))
```

```python
# 3d Visualize

fig = plt.figure(figsize = (15, 12))
ax = fig.add_subplot(111, projection="3d")
ax.grid(True, color="#323232")
ax.set_facecolor("black")
ax.figure.set_facecolor("#121212")
ax.tick_params(axis="x", colors="white")
ax.tick_params(axis="y", colors="white")


for point in points["blue"]:
  ax.scatter(point[0], point[1], point[2], color="#104DCA", s=60)
for point in points["red"]:
  ax.scatter(point[0], point[1], point[2],color="#FF0000", s=60)



new_class = clf.predict(new_points)
color = "#FF0000" if new_class == "red" else "#104DCA"
ax.scatter(new_points[0], new_points[1], new_points[2], color=color, marker="*", s=200, zorder=100)


for point in points["blue"]:
  ax.plot([point[0], new_points[0]], [point[1], new_points[1]], [point[2], new_points[2]] , color="#104DCA", linestyle="--")
for point in points["red"]:
  ax.plot([point[0], new_points[0]], [point[1], new_points[1]], [point[2], new_points[2]], color="#FF0000", linestyle="--")

plt.show()
```
